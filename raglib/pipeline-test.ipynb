```python
from raglib import RAG
from raglib.models import (
    GARAGIndexerConfig,
    GraphRAGIndexerConfig,
    NaiveGRIndexerConfig,
    NaiveRAGIndexerConfig,
    GARAGRetrieverConfig,
    GraphRAGRetrieverConfig,
    NaiveGRRetrieverConfig,
    NaiveRAGRetrieverConfig
)
from raglib.graphrag import (
    GARAGIndexer, 
    GraphRAGIndexer,
    NaiveGraphRAGIndexer,
    NaiveRAGIndexer,
    GARAGRetriever,
    GraphRAGRetriever,
    NaiveGraphRAGRetriever,
    NaiveRAGRetriever
)
```

#### Custom Configuration

```python
indexer_config_dict = {"max_chunk_size": 1234}
indexer_config_class = GARAGIndexerConfig(**indexer_config_dict)

indexer_dict = GARAGIndexer(indexer_config_dict) # parameter=indexer_config_dict
indexer_class = GARAGIndexer(config=indexer_config_class)
```

```python
print(indexer_config_class)
print(indexer_dict.config)
print(indexer_class.config)
```

```python
retriever_config_dict = {"top_k": 1000, "similarity_metric": "l2"}
retriever_config_class = GARAGRetrieverConfig(**retriever_config_dict)

retriever_dict = GARAGRetriever(retriever_config_dict) # parameter=retriever_config_dict
retriever_class = GARAGRetriever(config=retriever_config_class)
```

```python
print(retriever_config_class)
print(retriever_dict.config)
print(retriever_class.config)
```

#### Instantiation

```python
test = RAG(indexer= "graphrag", retriever="graphrag", verbosity=1)
```

```python
test = RAG(indexer="graphrag", indexer_config=indexer_config_dict, retriever="graphrag", retriever_config=retriever_config_dict)
```

```python
test = RAG(indexer="garag", indexer_config=indexer_config_class, retriever="garag", retriever_config=retriever_config_class)
```

```python
test = RAG(indexer=indexer_dict, retriever=retriever_dict)
```

```python
test = RAG(indexer=indexer_class, retriever=retriever_class)
```

#### Indexing & Query Pipeline

**Indexing**

```python
test.index(verbosity=1)
```

**Retrieval**

```python
answer = test.query(prompt="What is the optimal chunk size for the indexing phase?", verbosity=1)
```

```python
for res in answer:
    print(res.matchedContent+"\n")
```
